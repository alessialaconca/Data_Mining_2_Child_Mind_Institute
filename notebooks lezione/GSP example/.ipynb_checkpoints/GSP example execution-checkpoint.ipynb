```python
import pandas as pd
import numpy as np

from gsp_python.gsp import load_ds
from gsp_python.gsp import GSP
from gsp_python.dataset_gen import DatasetGenerator
```

## Reading an input file

Method load_ds reads datasets with the same format [defined here](https://www.philippe-fournier-viger.com/spmf/GSP.php); the method converts all strings to integers, the first dictionary returned can be used to convert those integers back to strings

! Note that this implementation works even if provided with a list[list[list[string]]] dataset, conversion is done to ensure slightly more efficient comparisons during execution

```python
filename = './example_dataset.txt'
```

```python
dataset, reconv_dict, conv_dict = load_ds(input_filename=filename)
dataset
```

```python
for sequence in dataset:
    for element in sequence:
        for i in range(len(element)):
            element[i] = reconv_dict[element[i]]
dataset
```

## Generation of a random dataset

```python
# Generation of a small random sequential dataset
# 5 sequences, 3 distinct events, 3 events max per element, 3 elements on average per sequence
generator = DatasetGenerator(size=5, nevents=3, maxevents=3, avgelems=3, seed=42)
dataset = generator.generate_sequence_dataset()
```

```python
print(type(dataset))
dataset
```

## Simple execution example

```python
# Running GSP without time constraints

gsp = GSP(ds=dataset, minsup=0.5)
patterns = gsp.run_gsp()
patterns
```

```python
# Running GSP with time constraints

gsp = GSP(ds=dataset, minsup=0.5, mingap=1, maxgap=2, maxspan=3)
patterns = gsp.run_gsp()
patterns
```

Simple conversion to DataFrame

```python
def output_to_dataframe(patterns):
    df = pd.DataFrame(data=patterns, columns=['pattern', 'support_count'])
    return df
```

```python
output_to_dataframe(patterns)
```
